##  **Estabilidad y Respuesta en Frecuencia de Sistemas Realimentados de 2° Orden**


---

## 1. Introducción y Marco Teórico: Del Lazo Abierto al Lazo Cerrado

Al analizar la estabilidad y respuesta dinámica de un amplificador realimentado cuando el amplificador operacional (AO) no es ideal, recurrimos al modelo general de realimentación (ecuación de Blackman / ganancia de lazo).

En un sistema realimentado negativamente, la ganancia de lazo cerrado $A_{vf}(s)$ se relaciona con la ganancia directa de lazo abierto $A_v(s)$ y la ganancia de lazo $T(s)$ mediante la expresión fundamental:

$$A_{vf}(s) = \frac{A_v(s)}{1 - T(s)}$$

donde la ganancia de lazo se define como:

$$T(s) = -K \cdot A_d(s)$$

siendo $K$ el factor de realimentación (por ejemplo, $K = \frac{R_1}{R_1 + R_2}$ en una configuración no inversora con red resistiva pura) y $A_d(s)$ la ganancia diferencial en lazo abierto del amplificador operacional.

### 1.1. Modelo de Segundo Orden del Amplificador Operacional
Un amplificador operacional real posee múltiples polos. En el rango de frecuencias de interés para el análisis de estabilidad, su respuesta dinámica se modela de forma precisa mediante dos polos reales en lazo abierto ($\omega_1$ y $\omega_2$, donde $\omega_1 \ll \omega_2$ representa el polo dominante y el segundo polo respectivamente):

$$A_d(s) = \frac{A_{d0}}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)}$$

Por lo tanto, la ganancia de lazo $T(s)$ resulta:

$$T(s) = \frac{-K \cdot A_{d0}}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)} = \frac{-T_0}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)}$$

donde definimos la ganancia de lazo en continua como $T_0 = K \cdot A_{d0} > 0$.

---

### 1.2. Deducción de la Función de Transferencia de Lazo Cerrado y Parámetros ($\omega_p, Q_p$)

Reemplazando $A_d(s)$ y $T(s)$ en la expresión de lazo cerrado (asumiendo por simplicidad la configuración donde $A_v(s) = A_d(s)$):

$$A_{vf}(s) = \frac{\frac{A_{d0}}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)}}{1 + \frac{K A_{d0}}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)}} = \frac{A_{d0}}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right) + T_0}$$

Expandiendo el denominador:

$$D(s) = 1 + s\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) + \frac{s^2}{\omega_1 \omega_2} + T_0 = (1 + T_0) \left[ 1 + \frac{s}{1 + T_0}\left(\frac{\omega_1 + \omega_2}{\omega_1 \omega_2}\right) + \frac{s^2}{(1 + T_0)\omega_1 \omega_2} \right]$$

Dado que en cualquier aplicación práctica la ganancia de lazo en continua es muy elevada ($T_0 = K A_{d0} \gg 1$), podemos aproximar $1 + T_0 \approx T_0$. Así, la ganancia en continua de lazo cerrado tiende al valor ideal $A_{vfi} = \frac{A_{d0}}{1 + T_0} \approx \frac{1}{K}$, y la función de transferencia normalizada de lazo cerrado $a_f(s) = \frac{A_{vf}(s)}{A_{vf}(0)}$ adopta la forma canónica de un sistema de segundo orden:

$$a_f(s) = \frac{1}{1 + \frac{s}{\omega_p Q_p} + \frac{s^2}{\omega_p^2}}$$

Igualando término a término los coeficientes del polinomio denominador con la forma canónica estándar, obtenemos las definiciones fundamentales de la **frecuencia del polo ($\omega_p$)** y el **factor de calidad ($Q_p$)**:

1. **Término cuadrático ($\omega_p$):**
   $$\frac{1}{\omega_p^2} = \frac{1}{(1 + T_0)\omega_1 \omega_2} \approx \frac{1}{T_0 \omega_1 \omega_2} \implies \boxed{\omega_p = \sqrt{(1 + T_0)\omega_1 \omega_2} \approx \sqrt{T_0 \omega_1 \omega_2}}$$
   *(Se observa que $\omega_p$ es la media geométrica de los polos en lazo abierto escalada por la raíz de la ganancia de lazo $T_0$).*

2. **Término lineal ($Q_p$):**
   $$\frac{1}{\omega_p Q_p} = \frac{1}{1 + T_0}\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) \approx \frac{1}{T_0}\left(\frac{\omega_1 + \omega_2}{\omega_1 \omega_2}\right)$$
   Sustituyendo $\omega_p = \sqrt{T_0 \omega_1 \omega_2}$ y definiendo la relación de separación de polos en lazo abierto como $d_p = \frac{\omega_2}{\omega_1}$:
   $$\boxed{Q_p = \frac{\sqrt{T_0 \omega_1 \omega_2}}{\omega_1 + \omega_2} = \frac{\sqrt{T_0 \cdot d_p}}{d_p + 1}}$$
   Como habitualmente los polos están muy separados ($\omega_2 \gg \omega_1 \implies d_p \gg 1$), se cumple la útil aproximación de diseño:
   $$Q_p \approx \sqrt{\frac{T_0}{d_p}} = \sqrt{\frac{K A_{d0} \omega_1}{\omega_2}} = \sqrt{\frac{K \cdot \omega_T}{\omega_2}}$$
   donde $\omega_T = A_{d0} \omega_1$ es el producto ganancia-ancho de banda (GBW) del amplificador operacional.

---

### 1.3. Normalización en Frecuencia ($S = \frac{s}{\omega_p}$)

Para independizar el estudio matemático de las frecuencias particulares de cada circuito y obtener curvas universales de diseño que dependan exclusivamente del factor de calidad $Q_p$, definimos la variable compleja de frecuencia normalizada:

$$S = \frac{s}{\omega_p} \quad \xrightarrow{s = j\omega} \quad S = j\frac{\omega}{\omega_p} = j\Omega$$

donde $\Omega = \frac{\omega}{\omega_p}$ es la frecuencia angular normalizada respecto a $\omega_p$. Bajo esta transformación, la transferencia normalizada de lazo cerrado queda expresada únicamente en función de $Q_p$:

$$\boxed{a_f(S) = \frac{1}{S^2 + \frac{1}{Q_p}S + 1}}$$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# Configuración estética general para los gráficos del reporte
plt.rcParams['font.size'] = 11
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 13
plt.rcParams['legend.fontsize'] = 10
plt.rcParams['figure.dpi'] = 100

# Definimos el vector del factor de calidad Qp a evaluar
# Rango desde Qp = 0.1 (fuertemente sobreamortiguado) hasta Qp = 3.0 (fuertemente subamortiguado / con sobrepico)
Qp_values = np.linspace(0.1, 3.0, 600)

# Puntos notables de referencia en el diseño de filtros y amplificadores:
# 1. Máxima Planicidad de Módulo (MPM - Respuesta Butterworth): sin sobrepico en frecuencia
Qp_MPM = 1.0 / np.sqrt(2)  # ≈ 0.7071

# 2. Máxima Planicidad de Retardo de Grupo (MPR - Respuesta Bessel): fase lineal
Qp_MPR = 1.0 / np.sqrt(3)  # ≈ 0.5774

print(f"Qp para Máxima Planicidad de Módulo (MPM - Butterworth): {Qp_MPM:.4f}")
print(f"Qp para Máxima Planicidad de Retardo (MPR - Bessel):     {Qp_MPR:.4f}")
```

## 2. Deducción Analítica del Ancho de Banda Normalizado a $-3\text{ dB}$ ($\Omega_H$)

El ancho de banda a $-3\text{ dB}$ en lazo cerrado ($\omega_H$, o en su forma normalizada $\Omega_H = \frac{\omega_H}{\omega_p}$) es la frecuencia a la cual el módulo de la transferencia normalizada cae a $\frac{1}{\sqrt{2}}$ de su valor en continua (o equivalentemente, su módulo al cuadrado cae a $\frac{1}{2}$).

Evaluando $a_f(S)$ en régimen permanente sinusoidal ($S = j\Omega$):

$$a_f(j\Omega) = \frac{1}{(j\Omega)^2 + j\frac{\Omega}{Q_p} + 1} = \frac{1}{(1 - \Omega^2) + j\left(\frac{\Omega}{Q_p}\right)}$$

Calculamos el módulo al cuadrado $|a_f(j\Omega)|^2$ sumando los cuadrados de la parte real e imaginaria del denominador:

$$|a_f(j\Omega)|^2 = \frac{1}{(1 - \Omega^2)^2 + \left(\frac{\Omega}{Q_p}\right)^2}$$

Desarrollando el binomio $(1 - \Omega^2)^2 = 1 - 2\Omega^2 + \Omega^4$ y agrupando los términos de igual potencia en $\Omega$:

$$|a_f(j\Omega)|^2 = \frac{1}{1 - 2\Omega^2 + \Omega^4 + \frac{\Omega^2}{Q_p^2}} = \frac{1}{\Omega^4 + \left(\frac{1}{Q_p^2} - 2\right)\Omega^2 + 1}$$

Planteamos la condición de corte a $-3\text{ dB}$ en $\Omega = \Omega_H$:

$$|a_f(j\Omega_H)|^2 = \frac{1}{2} \implies \frac{1}{\Omega_H^4 + \left(\frac{1}{Q_p^2} - 2\right)\Omega_H^2 + 1} = \frac{1}{2}$$

Invirtiendo ambos miembros e igualando a cero:

$$\Omega_H^4 + \left(\frac{1}{Q_p^2} - 2\right)\Omega_H^2 + 1 = 2 \implies \boxed{\Omega_H^4 + \left(\frac{1}{Q_p^2} - 2\right)\Omega_H^2 - 1 = 0}$$

### Resolución de la Ecuación Bicuadrática
Esta expresión es una ecuación bicuadrática en $\Omega_H$. Realizando el cambio de variable auxiliar $u = \Omega_H^2$ (con la restricción física $u > 0$, ya que $\Omega_H$ es una frecuencia real positiva):

$$u^2 + b \cdot u - 1 = 0 \quad \text{donde} \quad b = \left(\frac{1}{Q_p^2} - 2\right)$$

Aplicando la fórmula resolvente de segundo grado ($a = 1$, $b = \frac{1}{Q_p^2} - 2$, $c = -1$):

$$u = \frac{-b \pm \sqrt{b^2 - 4(1)(-1)}}{2} = \frac{-b \pm \sqrt{b^2 + 4}}{2}$$

Dado que $\sqrt{b^2 + 4} > |b|$ para todo $b \in \mathbb{R}$, la única raíz real positiva ($u > 0$) corresponde al signo $+$. Finalmente, deshaciendo el cambio de variable ($\Omega_H = \sqrt{u}$), obtenemos la expresión analítica cerrada para el ancho de banda normalizado:

$$\boxed{\Omega_H(Q_p) = \sqrt{\frac{-\left(\frac{1}{Q_p^2} - 2\right) + \sqrt{\left(\frac{1}{Q_p^2} - 2\right)^2 + 4}}{2}}}$$

> **Caso particular notable (MPM - Butterworth):**  
> Si $Q_p = \frac{1}{\sqrt{2}} \approx 0.707 \implies Q_p^2 = \frac{1}{2} \implies \frac{1}{Q_p^2} - 2 = 2 - 2 = 0$.  
> El término cuadrático intermedio se anula exactamente (condición de máxima planicidad), quedando $\Omega_H^4 - 1 = 0 \implies \mathbf{\Omega_H = 1}$ (es decir, $\omega_H = \omega_p$).


```python
omega_H_values = []

for Qp in Qp_values:
    # Coeficiente b de la ecuación bicuadrática: u^2 + coef_b * u - 1 = 0
    coef_b = (1.0 / (Qp**2)) - 2.0
    
    # Raíz positiva de la cuadrática en u = Omega_H^2
    u = (-coef_b + np.sqrt(coef_b**2 + 4.0)) / 2.0
    
    # Ancho de banda normalizado Omega_H
    omega_H = np.sqrt(u)
    omega_H_values.append(omega_H)

omega_H_values = np.array(omega_H_values)

# Cálculo exacto en los puntos notables MPM y MPR
def calc_omega_H(Qp):
    b = (1.0 / (Qp**2)) - 2.0
    return np.sqrt((-b + np.sqrt(b**2 + 4.0)) / 2.0)

wH_MPM = calc_omega_H(Qp_MPM)
wH_MPR = calc_omega_H(Qp_MPR)

print(f"Para MPM (Qp = {Qp_MPM:.3f}) -> Omega_H = {wH_MPM:.4f}")
print(f"Para MPR (Qp = {Qp_MPR:.3f}) -> Omega_H = {wH_MPR:.4f}")
```

## 3. Ganancia de Lazo $T(S)$, Frecuencia de Cruce ($\Omega_G$) y Margen de Fase ($M\varphi$)

Para evaluar la estabilidad relativa del sistema mediante el criterio de Bode, necesitamos expresar la ganancia de lazo $T(s)$ en términos de la variable de frecuencia normalizada $S = \frac{s}{\omega_p}$ y del factor de calidad $Q_p$.

### 3.1. Deducción de la Expresión Normalizada de $T(S)$
Partimos de la relación fundamental entre la transferencia de lazo cerrado y la ganancia de lazo:

$$A_{vf}(s) = \frac{A_v(s)}{1 - T(s)} = \frac{A_{vfi}}{1 - \frac{1}{T(s)}}$$

En la Sección 1.2 demostramos que, para $T_0 \gg 1$, el denominador de lazo cerrado proviene directamente de desarrollar $1 - T(s)$:

$$1 - T(s) = 1 + \frac{T_0}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)} = \frac{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right) + T_0}{\left(1 + \frac{s}{\omega_1}\right)\left(1 + \frac{s}{\omega_2}\right)}$$

Observemos la estructura de $T(s)$ para frecuencias medias y altas ($\omega \gg \omega_1$, donde se ubica la zona de cruce de ganancia $\omega_G$). Desarrollando el denominador de $T(s)$:

$$T(s) = \frac{-T_0}{1 + s\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) + \frac{s^2}{\omega_1 \omega_2}} = \frac{-1}{\frac{1}{T_0} + \frac{s}{T_0}\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) + \frac{s^2}{T_0 \omega_1 \omega_2}}$$

Recordando las identificaciones exactas obtenidas en la Sección 1.2:
* $\omega_p^2 = T_0 \omega_1 \omega_2 \implies \frac{s^2}{T_0 \omega_1 \omega_2} = \left(\frac{s}{\omega_p}\right)^2 = S^2$
* $\frac{1}{\omega_p Q_p} = \frac{1}{T_0}\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) \implies \frac{s}{T_0}\left(\frac{1}{\omega_1} + \frac{1}{\omega_2}\right) = \frac{1}{Q_p}\left(\frac{s}{\omega_p}\right) = \frac{S}{Q_p}$
* Y despreciando el término $\frac{1}{T_0} \approx 0$ frente a los términos dinámicos en la región de cruce (ya que $T_0 \to \infty$, lo cual equivale a aproximar el primer polo $\omega_1$ como un integrador ideal en las cercanías de $\omega_G$):

Obtenemos la **expresión normalizada de la ganancia de lazo $T(S)$**:

$$\boxed{T(S) = \frac{-1}{S^2 + \frac{1}{Q_p}S} = \frac{-1}{S\left(S + \frac{1}{Q_p}\right)}}$$

---

### 3.2. Cálculo de la Frecuencia de Cruce de Ganancia Normalizada ($\Omega_G$)

La frecuencia de cruce de ganancia $\omega_G$ (o normalizada $\Omega_G = \frac{\omega_G}{\omega_p}$) se define como la frecuencia a la cual el módulo de la ganancia de lazo es unitario ($0\text{ dB}$):

$$|T(j\Omega_G)| = 1$$

Sustituyendo $S = j\Omega_G$ en la expresión de $T(S)$:

$$T(j\Omega_G) = \frac{-1}{(j\Omega_G)\left(j\Omega_G + \frac{1}{Q_p}\right)}$$

Tomando el módulo en ambos miembros:

$$|T(j\Omega_G)| = \frac{|-1|}{|j\Omega_G| \cdot \left|\frac{1}{Q_p} + j\Omega_G\right|} = \frac{1}{\Omega_G \sqrt{\Omega_G^2 + \frac{1}{Q_p^2}}} = 1$$

Elevando ambos miembros al cuadrado para eliminar la raíz:

$$\Omega_G^2 \left(\Omega_G^2 + \frac{1}{Q_p^2}\right) = 1 \implies \boxed{\Omega_G^4 + \frac{1}{Q_p^2}\Omega_G^2 - 1 = 0}$$

Nuevamente estamos frente a una ecuación bicuadrática. Haciendo la sustitución $v = \Omega_G^2$ (con $v > 0$):

$$v^2 + \left(\frac{1}{Q_p^2}\right)v - 1 = 0$$

Aplicando la fórmula resolvente cuadrática y tomando la única raíz positiva ($v > 0$), llegamos a la expresión cerrada de $\Omega_G$ en función de $Q_p$:

$$\boxed{\Omega_G(Q_p) = \sqrt{\frac{-\frac{1}{Q_p^2} + \sqrt{\left(\frac{1}{Q_p^2}\right)^2 + 4}}{2}}}$$

---

### 3.3. Deducción del Margen de Fase ($M\varphi$) y Obtención de la Curva Inversa

Expresamos $T(j\Omega)$ separando sus factores para hallar su argumento (fase) a la frecuencia de cruce $\Omega_G$:

$$T(j\Omega_G) = \frac{-1}{(j\Omega_G)\left(\frac{1}{Q_p} + j\Omega_G\right)} = \frac{-1 \cdot Q_p}{(j\Omega_G)\left(1 + j\Omega_G Q_p\right)}$$

Analizamos el aporte de fase de cada término:
1. El signo negativo del numerador ($-1$, propio de la realimentación negativa en la convención $1 - T(s)$) aporta $-180^\circ$ (o $+\pi$).
2. El polo en el origen $(j\Omega_G)$ en el denominador aporta $-90^\circ$ ($-\frac{\pi}{2}\text{ rad}$).
3. El segundo polo $(1 + j\Omega_G Q_p)$ en el denominador aporta $-\arctan(\Omega_G Q_p)$.

Por lo tanto, la fase total de la ganancia de lazo $\angle T(j\Omega_G)$ es:

$$\angle T(j\Omega_G) = -180^\circ - 90^\circ - \arctan(\Omega_G Q_p) = -270^\circ - \arctan(\Omega_G Q_p)$$

Dado que trabajamos con $T(s) = -K A_d(s)$ (donde el signo menos de la realimentación ya está incluido dentro de $T$), el punto crítico de inestabilidad ($1 - T(j\Omega) = 0 \implies T(j\Omega) = +1 = 1\angle -360^\circ$) se alcanza cuando la fase total llega a $-360^\circ$.

El **Margen de Fase ($M\varphi$)** se define entonces como la distancia angular desde la fase en $\Omega_G$ hasta el límite de inestabilidad ($-360^\circ$):

$$M\varphi = \angle T(j\Omega_G) - (-360^\circ) = \left[-270^\circ - \arctan(\Omega_G Q_p)\right] + 360^\circ$$

$$\boxed{M\varphi = 90^\circ - \arctan(\Omega_G \cdot Q_p)}$$

#### ¿Cómo graficar la relación inversa ($Q_p$ en función de $M\varphi$)?
Como se analizó en clase, dado que $\Omega_G$ depende a su vez de $Q_p$ mediante una raíz bicuadrática ($\Omega_G(Q_p)$), despejar analíticamente $Q_p$ de la ecuación $M\varphi = 90^\circ - \arctan(\Omega_G(Q_p) \cdot Q_p)$ o intentar resolver el sistema de forma directa invirtiendo la función resulta algebraica y numéricamente engorroso. 

Sin embargo, en Python podemos obtener la curva inversa de forma directa y exacta **invirtiendo los ejes al graficar** los vectores ya calculados (colocando el vector *margin_phase_values* en el eje de abscisas $x$ y el vector *Qp_values* en el eje de ordenadas $y$), lo que nos permite leer directamente qué $Q_p$ corresponde a cada Margen de Fase requerido.


```python
omega_G_values = []
margin_phase_values = []

for Qp in Qp_values:
    # 1. Cálculo de Omega_G (Frecuencia de cruce de ganancia |T(j*Omega_G)| = 1)
    # Ecuación bicuadrática: v^2 + coef_b_G * v - 1 = 0, con v = Omega_G^2
    coef_b_G = 1.0 / (Qp**2)
    v = (-coef_b_G + np.sqrt(coef_b_G**2 + 4.0)) / 2.0
    omega_G = np.sqrt(v)
    omega_G_values.append(omega_G)
    
    # 2. Cálculo del Margen de Fase (Mphi) en grados
    # Mphi = 90° - arctg(Omega_G * Qp)
    m_phi_rad = (np.pi / 2.0) - np.arctan(omega_G * Qp)
    m_phi_deg = np.degrees(m_phi_rad)
    margin_phase_values.append(m_phi_deg)

omega_G_values = np.array(omega_G_values)
margin_phase_values = np.array(margin_phase_values)

# Evaluamos en los puntos notables MPM y MPR
def calc_omega_G_and_Mphi(Qp):
    b_G = 1.0 / (Qp**2)
    wG = np.sqrt((-b_G + np.sqrt(b_G**2 + 4.0)) / 2.0)
    mphi = np.degrees((np.pi / 2.0) - np.arctan(wG * Qp))
    return wG, mphi

wG_MPM, Mphi_MPM = calc_omega_G_and_Mphi(Qp_MPM)
wG_MPR, Mphi_MPR = calc_omega_G_and_Mphi(Qp_MPR)
wG_Q1,  Mphi_Q1  = calc_omega_G_and_Mphi(1.0)

print(f"MPM (Butterworth, Qp = {Qp_MPM:.3f}): Omega_G = {wG_MPM:.4f}, Mphi = {Mphi_MPM:.2f}°")
print(f"MPR (Bessel,      Qp = {Qp_MPR:.3f}): Omega_G = {wG_MPR:.4f}, Mphi = {Mphi_MPR:.2f}°")
print(f"Caso Qp = 1.000:                   Omega_G = {wG_Q1:.4f}, Mphi = {Mphi_Q1:.2f}°")
```

## 5. Conclusiones e Interpretación Física de los Resultados

1. **Compromiso entre Ancho de Banda y Estabilidad ($\Omega_H$ vs. $M\varphi$):**  
   A medida que el factor de calidad $Q_p$ aumenta (por ejemplo, al aumentar la ganancia de lazo $T_0$ o al acercarse entre sí los polos de lazo abierto $\omega_1$ y $\omega_2$), el ancho de banda normalizado de lazo cerrado $\Omega_H$ se incrementa monotónicamente. Sin embargo, este ensanchamiento de banda se logra a expensas de una reducción directa en el Margen de Fase $M\varphi$, llevando al sistema hacia una respuesta subamortiguada con picos de resonancia en frecuencia (para $Q_p > 0.707$) y oscilaciones transitorias (*ringing*) en el dominio del tiempo.
2. **Condición de Butterworth (MPM, $Q_p = 1/\sqrt{2} \approx 0.707$):**  
   Constituye el punto de diseño óptimo cuando se busca maximizar el ancho de banda plano sin que aparezca sobrepico en la respuesta en frecuencia ($|a_f(j\Omega)| \le 1$ para todo $\Omega$). En esta condición se verifica exactamente que $\Omega_H = 1$ ($\omega_H = \omega_p$), la frecuencia de cruce de ganancia ocurre en $\Omega_G = 0.644$ y se garantiza un margen de fase robusto de $M\varphi = 65.5^\circ$.
3. **Condición de Bessel (MPR, $Q_p = 1/\sqrt{3} \approx 0.577$):**  
   Prioriza la linealidad de fase (retardo de grupo constante) para evitar la distorsión de forma de onda en señales compuestas o pulsos. Requiere un sistema más amortiguado ($M\varphi = 72.4^\circ$), lo cual reduce el ancho de banda a $-3\text{ dB}$ a un $78.6\%$ de la frecuencia natural $\omega_p$ ($\Omega_H = 0.786$).
4. **Utilidad de la Curva Inversa para Compensación:**  
   En problemas de síntesis y compensación de amplificadores operacionales, el diseñador suele partir de una especificación de estabilidad (un $M\varphi$ mínimo requerido). La gráfica inversa $Q_p(M\varphi)$ permite extraer de inmediato el $Q_p$ máximo admisible y, mediante la relación $Q_p \approx \sqrt{\frac{T_0}{d_p}}$, dimensionar la separación de polos necesaria ($d_p = \omega_2/\omega_1$) o la red de compensación correspondiente.
